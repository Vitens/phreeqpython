# Convert notebooks
Read notebooks and convert to html.

```python
from mkdocs_jupyter import nbconvert2
import os
```

```python
nbdir = "docs\\notebooks\\"
notebooks = [file for file in os.listdir(nbdir) if file.endswith("ipynb")]
```

```python
targetdir = "docs\\nb_html\\"
for notebook in notebooks:
    content = nbconvert2.nb2html(nb_path=nbdir + notebook)
    target = targetdir + notebook.replace(".ipynb", ".html") 
    with open(target, "w", encoding="utf-8") as f:
            f.write(content)   
```
